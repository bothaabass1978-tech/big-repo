```python
def perc_on_bar(data, z):
    '''
    plot
    feature: categorical feature
    the function won't work if a column is passed in hue parameter
    '''

    total = len(data[z])
    plt.figure(figsize=(15,5))
    ax = sns.countplot(data[z],palette='Paired',order = data[z].value_counts().index)
    for p in ax.patches:
        percentage = '{:.1f}%'.format(100 * p.get_height()/total)
        x = p.get_x() + p.get_width() / 2 - 0.05
        y = p.get_y() + p.get_height()

        ax.annotate(percentage, (x, y), size = 12)

    plt.show()
```

```python
from sklearn.cluster import KMeans

distortions = []
K = range(2, 10)
for k in K:
    kmeanModel = KMeans(n_clusters=k, random_state=4)
    kmeanModel.fit(data_pca)
    distortions.append(
        kmeanModel.inertia_
    )
```

```python
plt.figure(figsize=(8, 4))
plt.plot(K, distortions, "bx-")
plt.xlabel("k (Number of Clusters)")
plt.ylabel("Distortion (Inertia)")
plt.title("Elbow Method for Optimal k")
plt.show()
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

sil_score = []
cluster_list = range(3, 7)

for n_clusters in cluster_list:


    clusterer = KMeans(n_clusters=n_clusters, random_state=1)


    preds = clusterer.fit_predict(data_pca)


    score = silhouette_score(data_pca, preds)


    sil_score.append(score)


    print(
        "For n_clusters = {}, the silhouette score is {}".format(
            n_clusters, score
        )
    )
```

```python
from scipy.cluster.hierarchy import linkage, cophenet
from scipy.spatial.distance import pdist

distance_metrics = ["euclidean", "chebyshev", "mahalanobis", "cityblock"]


linkage_methods = ["single", "complete", "average"]

high_cophenet_corr = 0
high_dm_lm = [0, 0]

for dm in distance_metrics:
    for lm in linkage_methods:
        Z = linkage(
            data_pca, metric=dm, method=lm
        )
        c, coph_dists = cophenet(
            Z, pdist(data_pca)
        )
        print(
            "Cophenetic correlation for {} distance and {} linkage is {}さんもご検討ください。".format(
                dm.capitalize(), lm, c
            )
        )
        if (
            high_cophenet_corr < c
        ):
            high_cophenet_corr = (
                c
            )
            high_dm_lm[0] = dm
            high_dm_lm[1] = lm


print(
    f"\nHighest Cophenetic Correlation is {high_cophenet_corr} with {high_dm_lm[0].capitalize()} distance and {high_dm_lm[1]} linkage."
)
```

```python

print(
    "Highest cophenetic correlation is {}, which is obtained with {} distance and {} linkage.".format(
        high_cophenet_corr, high_dm_lm[0].capitalize(), high_dm_lm[1]
    )
)
```

```python
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import cophenet, dendrogram, linkage
from scipy.spatial.distance import pdist
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler
import pandas as pd
import os


file_path = "marketing_campaign(2).csv"
if not os.path.exists(file_path):
    raise FileNotFoundError(f"File '{file_path}' not found. Please upload it using 'from google.colab import files; uploaded = files.upload()' in a separate cell, or ensure it's in the correct directory.")
original_data = pd.read_csv(file_path)
data = original_data.copy()

data.drop("ID", axis=1, inplace=True)

data["Education"].replace("2n cycle", "Master", inplace=True)
data["Marital_Status"].replace(["Alone", "absurd", "YOLO"], "Single", inplace=True)


data.drop(index=[164, 655, 687, 1300, 1653, 2132, 2233], inplace=True)

data["Age"] = 2016 - pd.to_datetime(data["Year_Birth"], format="%Y").apply(lambda x: x.year)
data.drop(data[data["Age"] > 115].index, inplace=True)

data["Kids"] = data["Kidhome"] + data["Teenhome"]
data["Marital_Status"] = data["Marital_Status"].replace(["Married", "Together"], "Relationship")
data["Status"] = data["Marital_Status"].replace({"Single": 1, "Divorced": 1, "Widow": 1, "Absurd": 1, "Relationship": 2})
data["Family_Size"] = data["Status"] + data["Kids"]

data["Expenses"] = (
    data["MntWines"] +
    data["MntFruits"] +
    data["MntMeatProducts"] +
    data["MntFishProducts"] +
    data["MntSweetProducts"] +
    data["MntGoldProds"]
)

data["NumTotalPurchases"] = (
    data["NumDealsPurchases"] +
    data["NumWebPurchases"] +
    data["NumCatalogPurchases"] +
    data["NumStorePurchases"]
)

data["Dt_Customer"] = pd.to_datetime(data["Dt_Customer"], dayfirst=True, format='mixed')
data["day"] = "01-01-2015"
data["day"] = pd.to_datetime(data.day)
data["Engaged_in_days"] = (data["day"] - data["Dt_Customer"]).dt.days

data["TotalAcceptedCmp"] = (
    data["AcceptedCmp1"] +
    data["AcceptedCmp2"] +
    data["AcceptedCmp3"] +
    data["AcceptedCmp4"] +
    data["AcceptedCmp5"] +
    data["Response"]
)


data = data.drop(data[data["NumTotalPurchases"] == 0].index)
data['AmountPerPurchase'] = data['Expenses'] / data['NumTotalPurchases']

data["Income"] = data["Income"].fillna(data["Income"].median())

data_model = data.drop(
    columns=[
        "Year_Birth",
        "Dt_Customer",
        "Complain",
        "Response",
        "AcceptedCmp1",
        "AcceptedCmp2",
        "AcceptedCmp3",
        "AcceptedCmp4",
        "AcceptedCmp5",
        "Marital_Status",
        "Status",
        "Kids",
        "Education",
        "Kidhome",
        "Teenhome",
        "Income",
        "Age",
        "Family_Size",
        "day"
    ]
)


scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_model)

pca = PCA(n_components=2)
data_pca = pca.fit_transform(data_scaled)
# --------------------------------------------------

linkage_methods = ["single", "complete", "average"]
fig, axs = plt.subplots(len(linkage_methods), 1, figsize=(10, 12))

for i, method in enumerate(linkage_methods):
    Z = linkage(data_pca, metric="cityblock", method=method)

    dendrogram(
        Z,
        ax=axs[i],
        truncate_mode="level",
        p=5,
        show_leaf_counts=True,
    )

    axs[i].set_title(f"Dendrogram ({method.capitalize()} Linkage)")

    coph_corr, _ = cophenet(Z, pdist(data_pca, metric="cityblock"))

    axs[i].annotate(
        f"Cophenetic\nCorrelation\n{coph_corr:0.2f}",
        (0.80, 0.75),
        xycoords="axes fraction",
        fontsize=10,
        bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="black", lw=1),
    )

plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.cluster.hierarchy import cophenet, dendrogram, linkage
from scipy.spatial.distance import pdist
from sklearn.decomposition import PCA
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()
data_scaled = scaler.fit_transform(data_model)

pca = PCA(n_components=2)
data_pca = pca.fit_transform(data_scaled)


cov_matrix = np.cov(data_pca, rowvar=False)


linkage_methods = ["single", "complete", "average"]

fig, axs = plt.subplots(len(linkage_methods), 1, figsize=(10, 12))

for i, method in enumerate(linkage_methods):

    mahalanobis_distances = pdist(data_pca, metric="mahalanobis", VI=cov_matrix)


    Z = linkage(mahalanobis_distances, method=method, metric='precomputed')
    dendrogram(
        Z,
        ax=axs[i],
        truncate_mode="level",
        p=5,
        show_leaf_counts=True,
    )
    axs[i].set_title(f"Dendrogram ({method.capitalize()} Linkage - Mahalanobis)")

    # Cophenet correlation also needs to use the same precomputed distances
    coph_corr, _ = cophenet(Z, mahalanobis_distances)


    axs[i].annotate(
        f"Cophenetic\nCorrelation\n{coph_corr:0.2f}",
        (0.80, 0.75),
        xycoords="axes fraction",
        fontsize=10,
        bbox=dict(boxstyle="round,pad=0.3", fc="white", ec="black", lw=1),
    )

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
import os


file_path = 'marketing_campaign(2).csv'
if os.path.exists(file_path):
    data = pd.read_csv(file_path)


    data.drop("ID", axis=1, inplace=True, errors='ignore')
    data["Income"] = data["Income"].fillna(data["Income"].median())


    data_model = data.select_dtypes(include=['number'])


    scaler = StandardScaler()
    data_scaled = scaler.fit_transform(data_model)
    pca = PCA(n_components=2)
    data_pca = pd.DataFrame(pca.fit_transform(data_scaled))


    if 'HC_Labels' in data_pca.columns:
        data_pca.drop("HC_Labels", axis=1, inplace=True)

    if 'HC_Labels' in data.columns:
        data.drop("HC_Labels", axis=1, inplace=True)

    print("Labels dropped successfully.")
else:
    print(f"File '{file_path}' not found. Please ensure it is uploaded to the environment.")
```

```python

eps_value = [2,3]
min_sample_values = [6,20]


res = {eps_value[i]: min_sample_values for i in range(len(eps_value))}
```

```python
import numpy as np
from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score


high_silhouette_avg = -1
high_i_j = [0, 0]


for i in res.keys():
    z = res[i]
    for j in z:
        db = DBSCAN(eps=i, min_samples=j).fit(data_pca)
        core_samples_mask = np.zeros_like(db.labels_, dtype=bool)
        core_samples_mask[db.core_sample_indices_] = True
        labels = db.labels_


        if len(set(labels)) > 1:
            silhouette_avg = silhouette_score(data_pca, labels)
            print(
                "For eps value = " + str(i),
                "For min sample = " + str(j),
                "The average silhouette_score is :",
                silhouette_avg,
            )

            if high_silhouette_avg < silhouette_avg:
                high_silhouette_avg = silhouette_avg
                high_i_j[0] = i
                high_i_j[1] = j
        else:
            print(f"For eps={i}, min_samples={j}: Only noise detected, skipping score.")

print(f"\nBest params: eps={high_i_j[0]}, min_samples={high_i_j[1]} with score {high_silhouette_avg}")
```

#Conclusion and Recommendations
1. Comparison of Various Techniques and Relative Performance
To identify distinct customer segments, multiple clustering algorithms were evaluated on the reduced PCA feature space. Performance was assessed using Silhouette Score (measuring cluster cohesion and separation), interpretability of cluster boundaries, and domain alignment.
Algorithm	Silhouette Score	Business Interpretability	Key Finding / Performance
K-Means Clustering	Moderate - High	High	Produced clean, spherical clusters with balanced group sizes. Strong baseline but assumes equal variance.
Hierarchical Clustering (HC)	Moderate	High	Provided clear hierarchical relationships via dendrograms; effective for nested grouping but sensitive to noise.
DBSCAN	Low - Variable	Low	Successfully identified extreme outliers/noise (-1 label) but struggled with varying cluster densities across PCA components.
Gaussian Mixture Model (GMM)	Highest	High	Superior probabilistic cluster assignments; accounts for non-spherical shapes and overlapping segments effectively.
•	Relative Performance: Gaussian Mixture Model (GMM) performed relatively better than hard clustering methods (K-Means/HC) and density-based methods (DBSCAN). Because real-world customer behavior rarely falls into rigid, non-overlapping categories, GMM’s soft assignment approach captured overlapping preferences and produced the highest overall silhouette score with well-balanced cluster sizes.
•	Scope for Further Improvement:
o	Feature Engineering: Derive secondary ratio metrics (e.g., Spend per Visit, Digital vs. Store Ratio) prior to PCA to give algorithms stronger signals.
o	Alternative Dimensionality Reduction: Test non-linear techniques like t-SNE or UMAP for visualization alongside PCA.
o	Hyperparameter Tuning: Fine-tune GMM covariance types (full, tied, diag, spherical) and use Bayesian Information Criterion (BIC) curves to refine component counts.
2. Refined Insights
The analysis revealed 5 distinct customer segments within the dataset:
•	High-Value Power Users (Cluster 2): Represent the primary revenue drivers. They exhibit peak spending across core product categories and consistently high engagement across channels.
•	Mainstream / Core Customers (Cluster 1): The largest segment by volume. They demonstrate steady, predictable purchasing patterns and represent stable baseline revenue.
•	Deal & Promotion-Sensitive Segment (Cluster 3): Highly responsive to discount campaigns and catalog offers. Their overall spend is moderate, but heavily concentrated around promotional events.
•	Digital-First / Channel-Specific Segment (Cluster 4): Prefer online browsing and direct engagement channels over traditional physical stores.
•	Low-Engagement / Budget Segment (Cluster 0): Infrequent buyers with low average order values. Minimal cross-category engagement.
3. Proposal for the Final Solution Design
Proposed Model: Gaussian Mixture Model (GMM) with PCA Feature Input
Why GMM is the Best Solution to Adopt:
1.	Soft Assignment (Probabilistic Framework): Unlike K-Means, GMM assigns probability scores to each cluster for every customer. This allows marketing teams to identify "borderline" customers (e.g., a customer with 60% probability in Mainstream and 40% in High-Value) and target them with progression campaigns.
2.	Handles Varied Density & Geometry: Real-world transaction data often forms elliptical rather than spherical clusters. GMM accommodates varying shapes and cluster variances far better than K-Means or Hierarchical Clustering.
3.	Robust Separation: GMM maintains a high silhouette score while keeping cluster proportions balanced—avoiding the single large "catch-all" cluster often produced by DBSCAN or Agglomerative Clustering on PCA-transformed data.
4.	Actionable Business Deployment: The 5 GMM profiles map directly to tailored marketing strategies:
o	Cluster 2: Premium loyalty rewards and VIP retention.
o	Cluster 3: Targeted discount codes and seasonal promotional mailers.
o	Cluster 4: Website UI optimizations, email triggers, and digital-exclusive deals.
o	Cluster 0: Low-cost re-engagement / win-back email automations.
